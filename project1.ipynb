```python
from statistics import correlation

import pandas as pd
```

```python
df = pd.read_csv("NigeriaInflationRates.csv")
df
```

```python
df["Month"].unique()
```

```python
import calendar

df["Month"] = df["Month"].apply(lambda x: calendar.month_name[x])
df["Month"].unique()
```

```python
aver = df.groupby("Month")["Inflation_Rate"].mean()
aver
```

```python
average_inflation = df["Inflation_Rate"].mean()
round(float(average_inflation), 2)
```

```python
highest_inflation = df.groupby("Year")["Inflation_Rate"].mean().idxmax()
round(highest_inflation)
```

```python
monthly_inflation = df["Month"].mode().max()
round(monthly_inflation)
```

```python
correlation = df["Crude Oil Price"].corr(df["Inflation_Rate"])
float(round(correlation, 2))
```

```python
cpi_change = df[["CPI_Food",	"CPI_Energy",	"CPI_Health",	"CPI_Transport",	"CPI_Communication",	"CPI_Education"]].mean().idxmax()
cpi_change
```
